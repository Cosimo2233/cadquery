```python
import sys
import importlib
sys.path.append(r"D:\\workspace\\quarry_cad\\cadquery_models\\example")

import cadquery as cq
from jupyter_cadquery import show, set_defaults
import gear_generator
importlib.reload(gear_generator)
from gear_generator import SpurGearSpec, build_spur_gear

# 齿轮参数
spec = SpurGearSpec(
    module=1.5,
    tooth_count=12,
    thickness=8.0,
    pressure_angle_deg=20.0,
    bore_diameter=5.0,
    backlash=0.02,
)

print(spec.summary())

# 构建并显示模型
set_defaults(axes=True, grid=None, glass=False)
result = build_spur_gear(
    spec,
    involute_samples=18,
    root_arc_samples=8,
    tip_arc_samples=8,
)
show(result)
```
